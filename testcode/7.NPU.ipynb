```python
# 셀 1: YOLOv8n COCO 모델 실시간 추론
from ultralytics import YOLO
from picamera2 import Picamera2
from IPython.display import display, Image
import cv2, time, gc

model = YOLO("/home/tiki/Workspace/yolov8n_hailo_model")
with Picamera2() as camera:
    camera.configure(camera.create_preview_configuration(main={"size": (320, 180), "format": "RGB888"}, controls={"FrameRate": 5}))
    camera.start(); time.sleep(1)
    screen = display(None, display_id=True)
    try:
        while True:
            frame = camera.capture_array(); start = time.perf_counter()
            result = model(frame, verbose=False)[0]; ms = (time.perf_counter() - start) * 1000
            view = result.plot(); cv2.putText(view, f"Inference: {ms:.1f} ms", (5, 18), 0, .45, (0, 255, 0), 1)
            _, jpg = cv2.imencode(".jpg", view)
            screen.update(Image(data=jpg.tobytes()))
    except KeyboardInterrupt:
        print("추론 종료")
    finally:
        model.predictor = None; model = None; gc.collect()
```
